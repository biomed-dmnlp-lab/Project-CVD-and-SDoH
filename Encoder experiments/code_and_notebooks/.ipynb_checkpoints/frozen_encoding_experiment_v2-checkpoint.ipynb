# Helper Functions

```python
def get_file_names(directory):
    file_names = []
    # List all files in the directory
    for filename in os.listdir(directory):
        # Check if the current entry is a file
        if os.path.isfile(os.path.join(directory, filename)):
            file_names.append(os.path.join(directory, filename))
    return file_names
```

```python
def get_relationships(input_string):
  lines = input_string.split('\n')
  e_lines = [line for line in lines if line.startswith('E')]
  result = []
  for line in e_lines:
      pairs = line.split()
      buffer=[]
      buffer2={}
      for t in pairs[1:]:
        buffer.append(t.split(':')[1])
      for buff in buffer[1:]:
        #buffer2.append((buffer[0],buff))
        buffer2[buff]=buffer[0]
      result.append(buffer2)
  return result

def get_sdoh_data_dict(txt_files,ann_files):
  data_list=[]
  for txt,ann in zip(txt_files,ann_files):
    file=open(txt,'r')
    data_dict={}

#

    data_dict['clinical_note']=file.read()
    data_dict['filename']= re.search(r"\/mimic\/(.*?)\.txt",txt).groups()[0]
    file.close()
    entities, relations, attributes, groups = get_entities_relations_attributes_groups(ann)
    data_dict['entities']=entities
    data_dict['relations']=get_relationships(open(ann,'r').read())
    data_dict['attributes']=attributes
    data_dict['groups']=groups
    data_list.append(data_dict)
    new_dict = {}
    for i, dict_ in enumerate(data_dict['relations']):


      for k, v in dict_.items():
          if v in new_dict:
              new_dict[v].append(k)
          else:
              new_dict[v] = [k]
    data_dict['relations']=new_dict

      #####################################################
    event_type_dict={'Alcohol':'ALC',  'Drug':'DRG',  'Employment':'EMP',
        'LivingStatus':'LVS', 'Tobacco':'TOB'}


    new_relations=[]
    id_enti={}
    for enti in data_dict['entities'].values():
      if enti.type in list(event_type_dict.keys()):
        id_enti[enti.id]=enti.type

    result = [(id_enti[key], [key]+data_dict['relations'][key]) for key in id_enti]
    data_dict['event_rel']=result
    data_dict['enti_id_ignore']=list(id_enti.keys())

  return data_list
```

```python
def get_target_clincal(sample):
    clinical_text = sample['clinical_note']

    # Initialize the label dictionary for the multi-class multi-label output
    label = {
        'Alcohol': None,
        'Tobacco': None,
        'Drug': None,
        'LivingStatus': None,
        'Employment': None
    }

    # A helper function to find the value for a category in the attributes
    def get_attribute_value(category, target_ids, attribute_type):
        for attr_key, attribute in sample['attributes'].items():
            if attribute.target in target_ids and attribute.type == attribute_type:
                return attribute.values[0]  # Assuming only one value per attribute
        return None

    # Process each category in the event_rel
    for category, target_ids in sample['event_rel']:
        if category == 'Tobacco':
            # Check if Tobacco is in the event_rel and get its attribute value
            label['Tobacco'] = get_attribute_value('Tobacco', target_ids, 'StatusTimeVal')
        elif category == 'Alcohol':
            # Check if Alcohol is in the event_rel and get its attribute value
            label['Alcohol'] = get_attribute_value('Alcohol', target_ids, 'StatusTimeVal')
        elif category == 'Drug':
            # Check if Drug is in the event_rel and get its attribute value
            label['Drug'] = get_attribute_value('Drug', target_ids, 'StatusTimeVal')
        elif category == 'LivingStatus':
            # LivingStatus can have two different types, StatusTimeVal and TypeLivingVal
            # First check for StatusTimeVal (current, past, future)
            status_value = get_attribute_value('LivingStatus', target_ids, 'StatusTimeVal')
            # Then check for TypeLivingVal (alone, with family, etc.)
            type_value = get_attribute_value('LivingStatus', target_ids, 'TypeLivingVal')

            if status_value and type_value:
                # Multiply the values by concatenating them as a string (you can change the format here if needed)
                label['LivingStatus'] = f'{status_value}_{type_value}'
            else:
                # If one or both are missing, return None
                label['LivingStatus'] = None
        elif category == 'Employment':
            # Employment has values like employed, unemployed, retired, etc.
            label['Employment'] = get_attribute_value('Employment', target_ids, 'StatusEmployVal')


    # Define all possible values for each label
    alcohol_values = ['none', 'current', 'past']
    tobacco_values = ['none', 'current', 'past']
    drug_values = ['none', 'current', 'past']
    living_status_values = [
        'current_with_family', 'current_alone', 'current_with_others', 'current_homeless',
        'past_with_family', 'past_alone', 'past_with_others', 'past_homeless',
        'future_with_family', 'future_alone', 'future_with_others', 'future_homeless'
    ]
    employment_values = ['employed', 'unemployed', 'retired', 'on_disability', 'student', 'homemaker']


    # Function to return the index of the value in the list or -1 if None
    def get_value_index(value, value_list):
        if value is None:
            return 0  
        else:
            try:
                return value_list.index(value)+1  # Return the index of the value
            except ValueError:
                return 0  # In case the value is not found (shouldn't happen if data is correct)

    # Getting the numerical representation (index) of the labels
    numerical_labels = {
        'Alcohol': get_value_index(label['Alcohol'], alcohol_values),
        'Tobacco': get_value_index(label['Tobacco'], tobacco_values),
        'Drug': get_value_index(label['Drug'], drug_values),
        # 'LivingStatus': get_value_index(label['LivingStatus'], living_status_values),
        'Employment': get_value_index(label['Employment'], employment_values)
    }

    # Return the clinical_text along with the labels
    return clinical_text, numerical_labels
```

```python
# Metrics for grouped multiclass outputs
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score


def build_compute_metrics(head_sizes):
    head_sizes = dict(head_sizes)
    head_names = list(head_sizes)

    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        if isinstance(logits, tuple):
            logits = logits[0]

        logits = np.asarray(logits)
        labels = np.asarray(labels, dtype=np.int64)

        expected_width = sum(head_sizes.values())
        if logits.ndim != 2 or logits.shape[1] != expected_width:
            raise ValueError(
                f"Expected logits [n_samples, {expected_width}], got {logits.shape}."
            )
        if labels.ndim != 2 or labels.shape[1] != len(head_sizes):
            raise ValueError(
                f"Expected labels [n_samples, {len(head_sizes)}], got {labels.shape}."
            )

        metrics = {}
        predictions_per_head = []
        valid_masks = []
        start_idx = 0

        for head_idx, (head_name, n_classes) in enumerate(head_sizes.items()):
            end_idx = start_idx + n_classes
            head_logits = logits[:, start_idx:end_idx]
            head_labels = labels[:, head_idx]
            valid = head_labels != -100

            if not valid.any():
                metrics[f"{head_name}_accuracy"] = float("nan")
                metrics[f"{head_name}_f1_macro"] = float("nan")
                metrics[f"{head_name}_f1_weighted"] = float("nan")
                metrics[f"{head_name}_roc_auc_ovr_macro"] = float("nan")
                predictions_per_head.append(np.full(len(labels), -100))
                valid_masks.append(valid)
                start_idx = end_idx
                continue

            head_predictions = np.argmax(head_logits, axis=1)
            predictions_per_head.append(head_predictions)
            valid_masks.append(valid)

            y_true = head_labels[valid]
            y_pred = head_predictions[valid]
            metrics[f"{head_name}_accuracy"] = accuracy_score(y_true, y_pred)
            metrics[f"{head_name}_f1_macro"] = f1_score(
                y_true,
                y_pred,
                labels=np.arange(n_classes),
                average="macro",
                zero_division=0,
            )
            metrics[f"{head_name}_f1_weighted"] = f1_score(
                y_true,
                y_pred,
                labels=np.arange(n_classes),
                average="weighted",
                zero_division=0,
            )

            shifted = head_logits[valid] - head_logits[valid].max(
                axis=1, keepdims=True
            )
            exp_logits = np.exp(shifted)
            probabilities = exp_logits / exp_logits.sum(axis=1, keepdims=True)
            try:
                metrics[f"{head_name}_roc_auc_ovr_macro"] = roc_auc_score(
                    y_true,
                    probabilities,
                    labels=np.arange(n_classes),
                    multi_class="ovr",
                    average="macro",
                )
            except ValueError:
                metrics[f"{head_name}_roc_auc_ovr_macro"] = float("nan")

            start_idx = end_idx

        predictions = np.column_stack(predictions_per_head)
        valid_matrix = np.column_stack(valid_masks)
        fully_labeled = valid_matrix.all(axis=1)
        metrics["exact_match_accuracy"] = (
            float((predictions[fully_labeled] == labels[fully_labeled]).all(axis=1).mean())
            if fully_labeled.any()
            else float("nan")
        )

        metrics["mean_head_f1_macro"] = float(
            np.nanmean([
                metrics[f"{head_name}_f1_macro"]
                for head_name in head_names
            ])
        )
        metrics["mean_head_accuracy"] = float(
            np.nanmean([
                metrics[f"{head_name}_accuracy"]
                for head_name in head_names
            ])
        )
        return metrics

    return compute_metrics
```

# Model

```python
from transformers.modeling_outputs import SequenceClassifierOutput


class MultiHeadSequenceClassifier(nn.Module):
    """One shared token encoder and one multiclass head per target."""

    def __init__(
        self,
        model_dir,
        head_sizes,
        local_files_only=True,
        dropout=None,
    ):
        super().__init__()
        self.head_sizes = dict(head_sizes)
        self.head_names = list(self.head_sizes)
        self.encoder_is_frozen = False

        self.config = AutoConfig.from_pretrained(
            model_dir,
            local_files_only=local_files_only,
        )
        self.encoder = AutoModel.from_pretrained(
            model_dir,
            config=self.config,
            local_files_only=local_files_only,
        )

        hidden_size = self.config.hidden_size
        if dropout is None:
            dropout = getattr(self.config, "classifier_dropout", None)
        if dropout is None:
            dropout = getattr(self.config, "hidden_dropout_prob", 0.1)

        self.dropout = nn.Dropout(dropout)
        self.heads = nn.ModuleDict({
            head_name: nn.Linear(hidden_size, n_classes)
            for head_name, n_classes in self.head_sizes.items()
        })

        ################# bloack for token embeddings -> logit #################
        self.token_projection = nn.Sequential(
            nn.Linear(hidden_size, config.token_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )

        self.tokenEmb_heads = nn.ModuleDict({
            head_name: nn.Linear(config.max_length * config.token_dim, n_classes)
            for head_name, n_classes in self.head_sizes.items()
        })

    def freeze_encoder(self):
        for parameter in self.encoder.parameters():
            parameter.requires_grad = False
        self.encoder_is_frozen = True
        self.encoder.eval()

    def train(self, mode=True):
        super().train(mode)
        if self.encoder_is_frozen:
            self.encoder.eval()
        return self

    def masked_mean_pooling(self, token_embeddings, attention_mask):
        mask = attention_mask.unsqueeze(-1).float()
        summed = (token_embeddings*mask).sum(dim=1)
        counts = mask.sum(dim=1).clamp(min=1e-9)         # (B, 1)
        return summed / counts


    def forward(
        self,
        input_ids=None,
        attention_mask=None,
        token_type_ids=None,
        labels=None,
        **kwargs,
    ):
        encoder_inputs = {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            **kwargs,
        }
        if token_type_ids is not None:
            encoder_inputs["token_type_ids"] = token_type_ids

        encoder_outputs = self.encoder(**encoder_inputs)

        if not config.tokens_embed_to_logits:
        
            if not config.mean_pooling:
                pooled_output = self.dropout(
                    encoder_outputs.last_hidden_state[:, 0] # clas token
                )
            else:
                pooled_output = self.dropout(self.masked_mean_pooling(encoder_outputs.last_hidden_state, attention_mask))
            logits = torch.cat(
                [
                    self.heads[head_name](pooled_output)
                    for head_name in self.head_names
                ],
                dim=-1,
            )
        else:
            x = self.token_projection(encoder_outputs.last_hidden_state)
            x = x * attention_mask.unsqueeze(-1).float()
            x = x.flatten(start_dim = 1)
            
            logits = torch.cat(
                [
                    self.tokenEmb_heads[head_name](x)
                    for head_name in self.head_names
                ],
                dim=-1,
            )
            
        return SequenceClassifierOutput(loss=None, logits=logits)


class MultiHeadEmbeddingClassifier(nn.Module):
    """Classification heads trained on frozen, precomputed embeddings."""

    def __init__(self, embedding_dim, head_sizes, dropout=0.1):
        super().__init__()
        self.embedding_dim = int(embedding_dim)
        self.head_sizes = dict(head_sizes)
        self.head_names = list(self.head_sizes)
        self.dropout = nn.Dropout(dropout)
        self.heads = nn.ModuleDict({
            head_name: nn.Linear(self.embedding_dim, n_classes)
            for head_name, n_classes in self.head_sizes.items()
        })

    def forward(self, embeddings=None, labels=None, **kwargs):
        if embeddings is None:
            raise ValueError("The Qwen classifier requires an 'embeddings' tensor.")
        if embeddings.ndim != 2 or embeddings.shape[1] != self.embedding_dim:
            raise ValueError(
                f"Expected embeddings [batch, {self.embedding_dim}], "
                f"got {tuple(embeddings.shape)}."
            )

        features = self.dropout(embeddings.float())
        logits = torch.cat(
            [
                self.heads[head_name](features)
                for head_name in self.head_names
            ],
            dim=-1,
        )
        return SequenceClassifierOutput(loss=None, logits=logits)
```

## Backbone options

Set `config.backbone_type` to one of:

- `"qwen3_vl_embedding"`: loads the local `Qwen3-VL-Embedding-2B` directory with Sentence Transformers, encodes each clinical note once, caches the normalized 2048-dimensional embeddings, unloads the 2B model, and trains only the multiclass heads.
- `"transformer_encoder"`: uses the original local `AutoModel` token encoder and freezes it while training the heads.

Qwen model documentation: https://huggingface.co/Qwen/Qwen3-VL-Embedding-2B

The Qwen path requires `sentence-transformers`. The model card's lower-level implementation lists `transformers>=4.57.0`, `qwen-vl-utils>=0.0.14`, and `torch==2.8.0`. This notebook uses the higher-level Sentence Transformers API and local-only loading.

# Dataset and Data Loaders

```python
import hashlib


HEAD_SIZES = {
    "Alcohol": 4,
    "Tobacco": 4,
    "Drug": 4,
    "Employment": 7,
}

CLASS_NAMES = {
    "Alcohol": ["missing_or_unknown", "none", "current", "past"],
    "Tobacco": ["missing_or_unknown", "none", "current", "past"],
    "Drug": ["missing_or_unknown", "none", "current", "past"],
    "Employment": [
        "missing_or_unknown",
        "employed",
        "unemployed",
        "retired",
        "on_disability",
        "student",
        "homemaker",
    ],
}

TASK_WEIGHTS = {head_name: 1.0 for head_name in HEAD_SIZES}
HEAD_NAMES = list(HEAD_SIZES)
N_HEADS = len(HEAD_SIZES)
TOTAL_LOGITS = sum(HEAD_SIZES.values())


def labels_to_tensor(label_dict):
    return torch.tensor(
        [label_dict[column] for column in HEAD_NAMES],
        dtype=torch.long,
    )


class ClinicalNotesDataset(Dataset):
    """Tokenized text dataset for the AutoModel backbone."""

    def __init__(self, texts, labels, tokenizer, max_length=512):
        if len(texts) != len(labels):
            raise ValueError(
                f"Length mismatch: {len(texts)} texts vs {len(labels)} labels."
            )
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            str(self.texts[idx]),
            add_special_tokens=True,
            max_length=self.max_length,
            padding="max_length",
            truncation=True,
            return_attention_mask=True,
            return_tensors="pt",
        )
        item = {
            key: value.squeeze(0)
            for key, value in encoding.items()
        }
        item["labels"] = labels_to_tensor(self.labels[idx])
        return item


class PrecomputedEmbeddingDataset(Dataset):
    """Qwen embedding vectors paired with grouped multiclass labels."""

    def __init__(self, embeddings, labels):
        self.embeddings = torch.as_tensor(embeddings).detach().cpu().float()
        self.labels = labels
        if len(self.embeddings) != len(self.labels):
            raise ValueError(
                f"Length mismatch: {len(self.embeddings)} embeddings vs "
                f"{len(self.labels)} labels."
            )
        if self.embeddings.ndim != 2:
            raise ValueError(
                f"Expected a 2D embedding matrix, got {self.embeddings.shape}."
            )

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            "embeddings": self.embeddings[idx],
            "labels": labels_to_tensor(self.labels[idx]),
        }


def _embedding_fingerprint(texts, model_dir, instruction, max_length):
    digest = hashlib.sha256()
    digest.update(str(Path(model_dir).resolve()).encode("utf-8"))
    digest.update(instruction.encode("utf-8"))
    digest.update(str(max_length).encode("utf-8"))
    for text in texts:
        encoded = str(text).encode("utf-8")
        digest.update(len(encoded).to_bytes(8, "little"))
        digest.update(encoded)
    return digest.hexdigest()[:16]


def encode_or_load_qwen_embeddings(
    embedder,
    texts,
    split_name,
    model_dir,
):
    # cache_dir = Path(config.embedding_cache_directory)
    # cache_dir.mkdir(parents=True, exist_ok=True)
    # fingerprint = _embedding_fingerprint(
    #     texts,
    #     model_dir=model_dir,
    #     instruction=config.qwen_instruction,
    #     max_length=config.max_length,
    # )
    # cache_path = cache_dir / f"{split_name}_{fingerprint}.pt"

    # if cache_path.exists():
    #     embeddings = torch.load(cache_path, map_location="cpu")
    #     print(f"Loaded cached {split_name} embeddings: {cache_path}")
    #     return embeddings.float()

    embeddings = embedder.encode(
        [str(text) for text in texts],
        prompt=config.qwen_instruction,
        batch_size=config.qwen_embedding_batch_size,
        show_progress_bar=True,
        convert_to_tensor=True,
        normalize_embeddings=True,
    )
    embeddings = embeddings.detach().cpu().float()
    # torch.save(embeddings, cache_path)
    # print(
    #     f"Created {split_name} embeddings with shape "
    #     f"{tuple(embeddings.shape)}: {cache_path}"
    # )
    return embeddings


if config.backbone_type == "transformer_decoder":
    try:
        from sentence_transformers import SentenceTransformer
    except ImportError as exc:
        raise ImportError(
            "Qwen embedding mode requires sentence-transformers. "
            "Install it in this environment before running the notebook."
        ) from exc

    model_dir = str(
        Path(config.model_directory) / config.qwen_embedding_model_name
    )
    if not Path(model_dir).is_dir():
        raise FileNotFoundError(
            f"Local Qwen embedding directory not found: {model_dir}"
        )

    qwen_device = "cuda" if torch.cuda.is_available() else "cpu"
    qwen_model_kwargs = {}
    if qwen_device == "cuda":
        qwen_model_kwargs["torch_dtype"] = torch.float16

    embedding_model = SentenceTransformer(
        model_dir,
        device=qwen_device,
        local_files_only=True,
        trust_remote_code=True,
        model_kwargs=qwen_model_kwargs,
    )
    embedding_model.max_seq_length = config.max_length

    train_embeddings = encode_or_load_qwen_embeddings(
        embedding_model,
        clinical_notes_train,
        "train",
        model_dir,
    )
    dev_embeddings = encode_or_load_qwen_embeddings(
        embedding_model,
        clinical_notes_dev,
        "dev",
        model_dir,
    )
    test_embeddings = encode_or_load_qwen_embeddings(
        embedding_model,
        clinical_notes_test,
        "test",
        model_dir,
    )

    embedding_dim = train_embeddings.shape[1]
    if dev_embeddings.shape[1] != embedding_dim:
        raise ValueError("Train and development embedding dimensions differ.")
    if test_embeddings.shape[1] != embedding_dim:
        raise ValueError("Train and test embedding dimensions differ.")

    train_dataset = PrecomputedEmbeddingDataset(
        train_embeddings,
        labels_train,
    )
    dev_dataset = PrecomputedEmbeddingDataset(
        dev_embeddings,
        labels_dev,
    )
    test_dataset = PrecomputedEmbeddingDataset(
        test_embeddings,
        labels_test,
    )
    model = MultiHeadEmbeddingClassifier(
        embedding_dim=embedding_dim,
        head_sizes=HEAD_SIZES,
    )

    # The frozen 2B embedder is no longer needed during head training.
    del embedding_model
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

elif config.backbone_type == "transformer_encoder":
    model_dir = str(
        Path(config.model_directory) / config.transformer_model_name
    )
    if not Path(model_dir).is_dir():
        raise FileNotFoundError(
            f"Local Transformer directory not found: {model_dir}"
        )

    tokenizer = AutoTokenizer.from_pretrained(
        model_dir,
        local_files_only=True,
    )
    model = MultiHeadSequenceClassifier(
        model_dir=model_dir,
        head_sizes=HEAD_SIZES,
        local_files_only=True,
    )
    train_dataset = ClinicalNotesDataset(
        clinical_notes_train,
        labels_train,
        tokenizer,
        max_length=config.max_length,
    )
    dev_dataset = ClinicalNotesDataset(
        clinical_notes_dev,
        labels_dev,
        tokenizer,
        max_length=config.max_length,
    )
    test_dataset = ClinicalNotesDataset(
        clinical_notes_test,
        labels_test,
        tokenizer,
        max_length=config.max_length,
    )
else:
    raise ValueError(
        "config.backbone_type must be 'qwen3_vl_embedding' "
        "or 'transformer_encoder'."
    )
```

```python
def validate_multiclass_labels(dataset, head_sizes, split_name):
    labels = np.asarray(
        [
            [row[head_name] for head_name in head_sizes]
            for row in dataset.labels
        ]
    )

    expected_shape = (len(dataset), len(head_sizes))
    if labels.shape != expected_shape:
        raise ValueError(
            f"{split_name}: expected labels shape {expected_shape}, "
            f"got {labels.shape}."
        )
    if not np.issubdtype(labels.dtype, np.integer):
        raise TypeError(
            f"{split_name}: labels must be integer class IDs, got {labels.dtype}."
        )

    print(f"\n{split_name}: labels shape={labels.shape}")
    for head_idx, (head_name, n_classes) in enumerate(head_sizes.items()):
        values = labels[:, head_idx]
        valid_values = values[values != -100]
        invalid = (valid_values < 0) | (valid_values >= n_classes)
        print(
            f"{head_name:12s} n_classes={n_classes} "
            f"unique={np.unique(valid_values).tolist()}"
        )
        if invalid.any():
            raise ValueError(
                f"{split_name}/{head_name}: invalid labels "
                f"{np.unique(valid_values[invalid]).tolist()}; expected "
                f"0..{n_classes - 1}, or -100 for an ignored target."
            )


validate_multiclass_labels(train_dataset, HEAD_SIZES, "TRAIN")
validate_multiclass_labels(dev_dataset, HEAD_SIZES, "DEV")
validate_multiclass_labels(test_dataset, HEAD_SIZES, "TEST")
```

# Training

```python
# Freeze the token backbone when using the original AutoModel route.
# Qwen vectors are already frozen and precomputed, so only heads exist.
if config.backbone_type == "transformer_encoder":
    model.freeze_encoder()

for parameter in model.heads.parameters():
    parameter.requires_grad = True

trainable_params = 0
frozen_params = 0
total_params = 0

for name, parameter in model.named_parameters():
    n_params = parameter.numel()
    total_params += n_params
    if parameter.requires_grad:
        trainable_params += n_params
        print(f"TRAINABLE: {name:70s} | {n_params:,}")
    else:
        frozen_params += n_params

print(f"\nBackbone mode: {config.backbone_type}")
print("Parameter summary")
print(
    f"Trainable: {trainable_params:,} / {total_params:,} "
    f"({100 * trainable_params / total_params:.4f}%)"
)
print(
    f"Frozen:    {frozen_params:,} / {total_params:,} "
    f"({100 * frozen_params / total_params:.4f}%)"
)

unexpected_trainable = [
    name
    for name, parameter in model.named_parameters()
    if parameter.requires_grad and not name.startswith("heads.")
]
if unexpected_trainable:
    raise RuntimeError(
        f"Unexpected trainable non-head parameters: {unexpected_trainable[:10]}"
    )
```

```python
with torch.no_grad():
    for parameter in model.parameters():
        if not parameter.is_contiguous():
            parameter.data = parameter.data.contiguous()

    for module in model.modules():
        for name, buffer in module.named_buffers(recurse=False):
            if buffer is not None and not buffer.is_contiguous():
                setattr(module, name, buffer.contiguous())
```

```python
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="steps",
    logging_steps=10,
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs= config.epochs,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="mean_head_f1_macro",
    greater_is_better=True,
    save_total_limit=2,
    report_to="none",
    # save_safetensors=False,
)

class MultiHeadTrainer(Trainer):
    """Weighted cross-entropy over multiple mutually exclusive targets."""

    def __init__(
        self,
        *args,
        head_sizes,
        task_weights=None,
        **kwargs,
    ):
        super().__init__(*args, **kwargs)
        self.head_sizes = dict(head_sizes)
        self.head_names = list(self.head_sizes)
        self.task_weights = (
            dict(task_weights)
            if task_weights is not None
            else {name: 1.0 for name in self.head_names}
        )

        if set(self.task_weights) != set(self.head_names):
            raise ValueError(
                "task_weights must contain exactly the HEAD_SIZES keys."
            )
        self.model_accepts_loss_kwargs = False

    def compute_loss(
        self,
        model,
        inputs,
        return_outputs=False,
        num_items_in_batch=None,
    ):
        labels = inputs["labels"].long()
        model_inputs = {
            key: value
            for key, value in inputs.items()
            if key != "labels"
        }
        outputs = model(**model_inputs)
        logits = outputs.logits

        if labels.ndim != 2 or labels.shape[1] != len(self.head_sizes):
            raise ValueError(
                f"Expected labels [batch, {len(self.head_sizes)}], "
                f"got {tuple(labels.shape)}."
            )

        expected_logits = sum(self.head_sizes.values())
        if logits.ndim != 2 or logits.shape[1] != expected_logits:
            raise ValueError(
                f"Expected logits [batch, {expected_logits}], "
                f"got {tuple(logits.shape)}."
            )

        weighted_losses = []
        active_weight = 0.0
        start_idx = 0

        for head_idx, (head_name, n_classes) in enumerate(
            self.head_sizes.items()
        ):
            end_idx = start_idx + n_classes
            head_logits = logits[:, start_idx:end_idx]
            head_labels = labels[:, head_idx]

            invalid = (
                (head_labels != -100)
                & ((head_labels < 0) | (head_labels >= n_classes))
            )
            if invalid.any().item():
                invalid_values = (
                    torch.unique(head_labels[invalid]).detach().cpu().tolist()
                )
                raise ValueError(
                    f"Head '{head_name}' received invalid labels "
                    f"{invalid_values}; expected 0..{n_classes - 1}, "
                    "or -100 to ignore."
                )

            if (head_labels != -100).any().item():
                weight = float(self.task_weights[head_name])
                head_loss = nn.functional.cross_entropy(
                    head_logits,
                    head_labels,
                    ignore_index=-100,
                )
                weighted_losses.append(weight * head_loss)
                active_weight += weight

            start_idx = end_idx

        if not weighted_losses:
            raise ValueError("Every target in this batch is ignored (-100).")

        loss = torch.stack(weighted_losses).sum() / active_weight
        return (loss, outputs) if return_outputs else loss


compute_metrics = build_compute_metrics(HEAD_SIZES)
```

```python
trainer = MultiHeadTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=dev_dataset,
    compute_metrics=compute_metrics,
    head_sizes=HEAD_SIZES,
    task_weights=TASK_WEIGHTS,
)

train_result = trainer.train()
print(train_result)

eval_results = trainer.evaluate()
print(eval_results)

test_results = trainer.evaluate(
    eval_dataset=test_dataset,
    metric_key_prefix="test",
)
print(test_results)

save_directory = (
    "./results/best_qwen_embedding_multihead_model"
    if config.backbone_type == "transformer_decoder"
    else "./results/best_frozen_transformer_multihead_model"
)
trainer.save_model(save_directory)
print(f"Saved classifier to {save_directory}")
```

```python
def predict_from_embedding_tensor(
    classifier,
    embeddings,
    head_sizes=HEAD_SIZES,
    class_names=CLASS_NAMES,
):
    device = next(classifier.parameters()).device
    classifier.eval()
    embeddings = torch.as_tensor(embeddings).float().to(device)

    with torch.no_grad():
        logits = classifier(embeddings=embeddings).logits

    predictions = {}
    start_idx = 0
    for head_name, n_classes in head_sizes.items():
        end_idx = start_idx + n_classes
        probabilities = torch.softmax(
            logits[:, start_idx:end_idx],
            dim=-1,
        ).cpu()
        class_ids = probabilities.argmax(dim=-1)
        predictions[head_name] = {
            "class_id": class_ids,
            "class_name": [
                class_names[head_name][class_id]
                for class_id in class_ids.tolist()
            ],
            "probabilities": probabilities,
        }
        start_idx = end_idx

    return predictions


def load_local_qwen_embedder():
    from sentence_transformers import SentenceTransformer

    local_model_dir = str(
        Path(config.model_directory) / config.qwen_embedding_model_name
    )
    device = "cuda" if torch.cuda.is_available() else "cpu"
    model_kwargs = (
        {"torch_dtype": torch.float16}
        if device == "cuda"
        else {}
    )
    embedder = SentenceTransformer(
        local_model_dir,
        device=device,
        local_files_only=True,
        trust_remote_code=True,
        model_kwargs=model_kwargs,
    )
    embedder.max_seq_length = config.max_length
    return embedder


def predict_clinical_notes_with_qwen(
    classifier,
    texts,
    embedder=None,
):
    if embedder is None:
        embedder = load_local_qwen_embedder()

    embeddings = embedder.encode(
        [str(text) for text in texts],
        prompt=config.qwen_instruction,
        batch_size=config.qwen_embedding_batch_size,
        show_progress_bar=False,
        convert_to_tensor=True,
        normalize_embeddings=True,
    )
    return predict_from_embedding_tensor(classifier, embeddings)


# Example:
# qwen_embedder = load_local_qwen_embedder()
# predictions = predict_clinical_notes_with_qwen(
#     model,
#     ["Patient currently smokes and is unemployed."],
#     embedder=qwen_embedder,
# )
# print(predictions["Tobacco"]["class_name"])
# print(predictions["Employment"]["probabilities"])
```
